# Per-Chapter Drills

A simpler, non-randomized version of `mock_interview.ipynb` for focused review of **one
chapter at a time** — every question in that chapter, in a fixed order, no sampling, no
branching. Use this after finishing a chapter, or the week before an interview to drill a
specific weak area. Same rule as everywhere else in this course: no model answers are ever
shown here — check `solutions/question_bank_answers.json` afterward.

## Setup

```python
import json
import sys
import time
from pathlib import Path

from IPython.core.error import StdinNotImplementedError

_repo_root = Path.cwd()
if not (_repo_root / "agentlib").is_dir():
    _repo_root = _repo_root.parent
if str(_repo_root) not in sys.path:
    sys.path.insert(0, str(_repo_root))

QUESTION_BANK_PATH = _repo_root / "interview_prep" / "question_bank.json"

with open(QUESTION_BANK_PATH) as f:
    question_bank = json.load(f)

# key_concepts lives with the answers in solutions/, not alongside the questions -- opening
# question_bank.json should not hand you the answer skeleton for all 93 of them. It is loaded
# here only to drive the post-answer keyword signal below, which you never see until after
# you have committed to an answer.
ANSWER_KEY_PATH = _repo_root / "solutions" / "question_bank_answers.json"

with open(ANSWER_KEY_PATH) as f:
    answer_key = json.load(f)


def key_concepts_for(question_id: str) -> list:
    return answer_key.get(question_id, {}).get("key_concepts", [])

print(f"Loaded {len(question_bank)} questions across 9 chapters.")
```

```python
def get_answer(prompt: str) -> str:
    '''Same headless-safe input() wrapper as mock_interview.ipynb -- catches both EOFError
    (plain Python, closed stdin) and IPython's StdinNotImplementedError (the ZMQ kernel used
    by nbconvert/nbmake, whose frontend doesn't implement stdin requests at all), so this
    notebook executes cleanly with zero errors under a headless test runner while still
    behaving like a normal input() prompt when run interactively.'''
    print(prompt)
    try:
        return input("Your answer: ")
    except (EOFError, StdinNotImplementedError):
        print("(no interactive input available in this run -- using a placeholder answer "
              "so the rest of this cell's logic still executes for real)")
        return "(placeholder answer -- no interactive input available)"


def keyword_feedback(answer_text: str, key_concepts: list) -> str:
    '''A rough, directional signal only -- not a grade.'''
    answer_lower = answer_text.lower()
    mentioned = [c for c in key_concepts if c.lower() in answer_lower]
    missed = [c for c in key_concepts if c.lower() not in answer_lower]
    lines = ["(rough keyword signal, not a grade -- see the solutions file for what actually matters)"]
    if mentioned:
        lines.append(f"  you mentioned: {', '.join(mentioned)}")
    if missed:
        lines.append(f"  you didn't mention: {', '.join(missed)}")
    return "\n".join(lines)
```

## Pick a chapter to drill

Set `CHAPTER` below to a number 1-9. Every question tagged with that chapter runs, in the
order it appears in `question_bank.json` — no sampling, no randomization.

```python
CHAPTER_NAMES = {
    1: "Fundamentals of AI Agents", 2: "Agent Control Flow", 3: "RAG and Retrieval Evaluation",
    4: "Production Reliability", 5: "Cost, Performance, and Model Selection",
    6: "Security and Safeguards", 7: "Tool Integration", 8: "System Design and Engineering Judgment",
    9: "LLMOps and Deployment",
}

CHAPTER = 6  # change this to drill a different chapter

drill_questions = [q for q in question_bank if q["chapter"] == CHAPTER]
print(f"Chapter {CHAPTER}: {CHAPTER_NAMES[CHAPTER]}")
print(f"{len(drill_questions)} questions queued, in bank order (no randomization).")
```

```python
def run_drill(questions: list) -> list:
    asked_ids = []
    for i, q in enumerate(questions, start=1):
        print(f"\n{'-' * 70}")
        print(f"Drill {i}/{len(questions)}  --  {q['difficulty']}, {q['category']}, format: {q['format']}")
        print("-" * 70)

        start = time.time()
        answer = get_answer(q["scenario"])
        elapsed = time.time() - start
        print(f"\n(elapsed: {elapsed:.0f}s -- soft timer, nothing was cut off)")

        print(keyword_feedback(answer, key_concepts_for(q["id"])))
        print(f"\nFollow-up: {q['follow_up']}")

        if q.get("grounded_in_real_incident"):
            print(f"(this scenario is grounded in a real public incident -- see "
                  f"question_bank.json's source_note for id {q['id']} if you want to read more)")

        asked_ids.append(q["id"])
    return asked_ids


asked = run_drill(drill_questions)
```

## Drill summary

As always: no model answers are shown here. Review against

```python
print(f"Drill complete -- Chapter {CHAPTER}: {CHAPTER_NAMES[CHAPTER]}")
print(f"{len(asked)} question(s) covered:")
for qid in asked:
    print(f"  - {qid}")
print(f"\nReview your answers against: solutions/question_bank_answers.json")
```
